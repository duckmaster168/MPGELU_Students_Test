Example: Calculating $\frac{dy}{dx}$ for $y = x^2 + 2x$ at $x = 3$

Solution:
$$
\dfrac{dy}{dx} = \dfrac{d}{dx}\left( x^2 + 2x\right)
$$
$$
= \dfrac{d}{dx}\left( x^2 \right) + \dfrac{d}{dx}\left(2x\right)
$$
$$
= 2x + 2
$$
At x = 3,
$$
y'(3) = 2(3) + 2 = 8
$$


```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2 + 2*x


dy_dx = torch.autograd.grad(outputs=y, inputs=x)[0]
print(f"The derivative of y with respect to x at x=3 is: {dy_dx.item()}")
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2 + 2*x
y.backward()
print(x.grad.item())
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2 + 2*x
y.backward()
x.grad.norm(2)
```

```python
w = torch.tensor([1.0, 2.0, -1.5], requires_grad=True)
# L = sum(w^2)
loss = (w ** 2).sum() 
print(f"Loss value: {loss.item():.4f}")
#Backward pass to populate w.grad
loss.backward()

# The gradient (dL/dw) is 2w = [2.0, 4.0, -3.0]
print(f"Gradient tensor: {w.grad}")

#Calculate the L2 norm of the gradient
#L2 norm = sqrt(2^2 + 4^2 + (-3)^2) = sqrt(29) ≈ 5.3852
grad_norm_l2 = torch.linalg.vector_norm(w.grad, ord=2) 


print(f"Gradient L2 norm: {grad_norm_l2.item():.4f}")
```
